```python

from pathlib import Path
from PIL import Image

# ==========================================================
# 1. FILE PATHS
# ==========================================================

input_file = Path(
    "/Users/bz294/Documents/WhiteZone/Figure/GEOSUS/"
    "Figure-R2/Figure_Final/Fig2.tiff"
)

output_file = input_file.with_name("Fig2_600dpi.tiff")

# ==========================================================
# 2. CONVERT TO 600 DPI
# ==========================================================

target_dpi = 600

if not input_file.exists():
    raise FileNotFoundError(f"File not found: {input_file}")

with Image.open(input_file) as img:

    print("Original file:", input_file)
    print("Original dimensions:", img.size)
    print("Original DPI:", img.info.get("dpi"))

    original_size = img.size

    # Change DPI metadata without resampling pixels
    img.save(
        output_file,
        format="TIFF",
        dpi=(target_dpi, target_dpi),
        compression="tiff_lzw"
    )

# ==========================================================
# 3. VERIFY OUTPUT
# ==========================================================

with Image.open(output_file) as img:

    new_dpi = img.info.get("dpi")
    new_size = img.size

    print("\n========== OUTPUT ==========")
    print("Saved to:", output_file)
    print("New DPI:", new_dpi)
    print("Pixel dimensions:", new_size)

    assert new_size == original_size, (
        "Pixel dimensions changed!"
    )

    assert new_dpi is not None and all(
        abs(d - target_dpi) < 0.1 for d in new_dpi
    ), "DPI verification failed!"

    width_cm = new_size[0] / target_dpi * 2.54
    height_cm = new_size[1] / target_dpi * 2.54

    print(f"Print size: {width_cm:.2f} × {height_cm:.2f} cm")
    print("SUCCESS: TIFF saved at 600 DPI.")
```
